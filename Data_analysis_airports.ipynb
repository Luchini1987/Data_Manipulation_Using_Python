```python
display(flights_datafr_raw)
```

```python
# The table above, it has 31 columns and 5.8 millions of rows.
# for our analysis, we are going to use the following dimensions : 
# ORIGIN_AIRPORT, DESTINATION_AIRPORT, YEAR, MONTH, DAY, AIRLINE, FLIGHT_NUMBER, SCHEDULED_DEPARTURE and DEPARTURE_DELAY.
flights_datafr_clean = flights_datafr_raw[['ORIGIN_AIRPORT', 'DESTINATION_AIRPORT','YEAR', 'MONTH', 'DAY', 'AIRLINE', 'FLIGHT_NUMBER',
'SCHEDULED_DEPARTURE', 'DEPARTURE_DELAY']]
display(flights_datafr_clean)
```

```python
# In order to ensure the quality of data, we are going to look for outliers, missing values or absurd values. 
flights_datafr_clean.describe()
```

```python
flights_datafr_clean.shape[0] - int(flights_datafr_clean['DEPARTURE_DELAY'].count())
# Looking at the table above, we can see that there are 86153 observations with null values. 
```

```python
# Next, we are going to remove these rows
flights_datafr_cleaned = flights_datafr_clean.dropna(axis= 0, how= 'any')
flights_datafr_cleaned.describe()
# The result after applying "dropna" is a table without null values and with 5732926 observations. 
```

```python
len((flights_datafr_cleaned['DESTINATION_AIRPORT'].unique()))
# Using the code above, we realize that there are 930 different destinations. For our analysis, we are only going to use
# the following :
# BOS, JFK, SFO AND LAX
```

```python
# The code below, it's used to filter the following dimensions:
# 'DESTINATION_AIRPORT'.- we are only going to analyse the airports ['BOS', 'JFK', 'SFO', 'LAX']
# 'DEPARTURE_DELAY' is a measure with a mean of 9.37 and a median of -2 . It means that there are outliers to the right and
# the values are right-skewed. We filter out values that have more than 1440 minutes of delay.

flights_datafr_cleaned = flights_datafr_cleaned[lambda x : (x['ORIGIN_AIRPORT'].isin(['BOS', 'JFK', 'SFO', 'LAX'])) & 
                                    (x['DEPARTURE_DELAY'] < 24*60)]
flights_datafr_cleaned.describe()
```

```python
# The result is a table with 535430 observations where the maximum value of the dimension 'DEPARTURE DELAY' is 1433 minutes
# which is correct. 
```

```python
# Before continuing the analysis, we need to create a new column which merges 4 columns YEAR, MONTH, DAY and SCHEDULED_DEPARTURE.

def convert_to_timedate(x, col='SCHEDULED_DEPARTURE'):
    'Converts to timestamp'
    return (
        pd.to_datetime(
            x['YEAR'].astype(str) + '-' + x['MONTH'].astype(str) + '-' + x['DAY'].astype(str) + 
            ' ' + x.assign(zero='0')['zero']*(4- x[col].astype(str).str.len()) + 
            x[col].astype(str), format= '%Y-%m-%d %H%M', errors = 'coerce'
        )
    )
```

```python
# The code below, it transforms the dimension 'FLIGHT_NUMBER' to the datatype 'string' and using the function 'convert_to_timedate',
# it merges 4 columns (YEAR, MONTH, DAY, SCHEDULED_DEPARTURE) and return a value DATETIME. 

flights_datafr_cleaned = flights_datafr_cleaned.assign(FLIGHT_NUMBER = lambda x: x['FLIGHT_NUMBER'].astype(str),
                             SCHEDULED_DEPARTURE = lambda x: convert_to_timedate(x),
                             is_delayed= lambda x: np.select(
                                 [x['DEPARTURE_DELAY'] >= 15], [1],
                                 default= 0
                             )
                        )
```

```python
display(flights_datafr_cleaned)
```

```python
# As we already used the columns YEAR, MONTH and DAY; These columns are deleted.
flights_datafr_cleaned = flights_datafr_cleaned.drop(['YEAR', 'MONTH', 'DAY'], axis= 1)
display(flights_datafr_cleaned)
```

```python
flights_datafr_cleaned.describe()
```

```python
# From the summary above, the dataset has been reduced to 535430 observations. 
# Also, the problem of the missing values has been solved and the column 'is_delayed' gives away that roughly 22% of
# flights have a delay. 
```

## Joining information from different sources

```python
# We want to obtain the name of the arrival airport from the CSV archive 'airports_datafr_raw'.
# The name of the airport has to be added to the dataframe 'flights_datafr_cleaned'.

# First, let's go to check out the dataframe 'airports_datafr_raw'.

airports_datafr_raw.head()
```

```python
flights_airports_merge = flights_datafr_cleaned.merge(airports_datafr_raw, left_on= 'ORIGIN_AIRPORT', right_on= 'IATA_CODE')
```

```python
# With the code above, 2 datasets have been joined using a column in commun called 'ORIGIN_AIRPORT' and 'IATA_CODE'.
```

```python
flights_airports_merge
```

## Calculating the number of flights per airport

```python
# The target is calculated the number of departing flights per airport across the year.
```

```python
flights_per_airport = flights_airports_merge.groupby(['ORIGIN_AIRPORT', 'AIRPORT']).apply(lambda x: pd.Series({'nb_flights' : len(x['FLIGHT_NUMBER'])})).reset_index()
display(flights_per_airport)
```

```python
flights_per_airport.assign(percentage_flights=lambda x: ((x['nb_flights'])/sum(x['nb_flights']))*100)
# The result shows that Los Angeles International Airport carries out the majority of flights with approximately 36 %
# followed by San Francisco International Airport with 27 %.
```

## Calculating the delayed flights for biggest airlines

```python
# The variable 'biggest airlines' is measured by the number of yearly flights.
```

```python
delayed_flights_big_airlines = flights_datafr_cleaned.groupby(['AIRLINE']).apply(lambda x: pd.Series({'nb_flights': len(x['FLIGHT_NUMBER']),
                                                                      'perc_delayed': round(np.average(x['is_delayed']), 4)
                                                                      })).reset_index()
display(delayed_flights_big_airlines)
# The result is a dataframe with the following results:
# The column 'nb_flights' has summarized the number de flights per airline.
# The column 'perc_delayed' presents the percentage of flights per airline. 
```

```python
# We need to range the airlines by the number of flights.
# With the code below , a new column has been created in order to obtain a ranking by the number of flights.

delayed_flights_big_airlines = delayed_flights_big_airlines.assign(rank_size= lambda x: x['nb_flights'].rank(method='first', ascending=False))
display(delayed_flights_big_airlines)
```

```python
delayed_flights_big_airlines = delayed_flights_big_airlines[lambda x: x['rank_size']<=10].drop('rank_size', axis = 1)
display(delayed_flights_big_airlines)

# The result of this code has kept the top ten.
# The column rank_size was deleted 
```

```python
# The column 'AIRLINE' only shows the code of each airline. 
# We need to add a new column with the full name of each airline.
# Finally, we need to range in descending order based on the column 'nb_flights'.

delayed_flights_big_airlines.merge(airlines_datafr_raw, left_on= 'AIRLINE', right_on= 'IATA_CODE').reset_index(drop=True).sort_values(
                                                                            by = ['perc_delayed'], ascending= False)
```

```python
# The result of the code above shows us that the airline 'Spirit Air Lines' has approximately 31% of delayed flights, 
# followed by the airline 'Southwest Airlines Co.' with 27% and at third place 'United Air Lines Inc.' with 23%.
# On the other hand, Alaska Airlines Inc., US Airways Inc. and American Airlines Inc. have between 12% and 15% of delayed flights.  
```

## Data Frame Transformations

```python
# We want to know the monthly change of flights by origin airport.
# 'flights_airports_merge' is the data frame that would be used.
```

```python
flights_airports_merge
```

```python
# From the column 'SCHEDULED_DEPARTURE', we are going to extract the month expressed like the full month name and 
# month number.
flights_airports_add_month= flights_airports_merge.assign(month=lambda x: x['SCHEDULED_DEPARTURE'].dt.strftime('%B'),
                              month_num=lambda x: x['SCHEDULED_DEPARTURE'].dt.strftime('%m').astype(int)
                             )
display(flights_airports_add_month)
```

```python
# Using the 'flights_airports_add_month' data frame obtained lines above, we are going to summarize the number of flights
# by 'ORIGIN_AIRPORT', 'month' and 'month_num'.
summarize_flights_airport_by_month = flights_airports_add_month.groupby(['ORIGIN_AIRPORT', 'month','month_num']).apply(lambda x: pd.Series(
                                    {'nb_flights': len(x['FLIGHT_NUMBER'])})).reset_index().sort_values(['ORIGIN_AIRPORT','month_num']
                                                                                                       ).reset_index(drop=True)
display(summarize_flights_airport_by_month)
```

```python
# Next, we want to compare the number of flights carries out over the current month against the month before.

summarize_flights_airport_by_month= summarize_flights_airport_by_month.assign(nb_flights_before= lambda x: x.groupby(['ORIGIN_AIRPORT'])['nb_flights'].shift(1),
                                          perc_chance=lambda x: round((x['nb_flights']- x['nb_flights_before'])/x['nb_flights_before'], 2)
                                         ).drop(['month_num'], axis = 1)
display(summarize_flights_airport_by_month)
```

## Most popular routes by origin airport

```python
# We are interested in knowing the top 3 routes per origin airport. 
# A top route from an airport is one with the highest number of flights departing from that airport.
# We use the data frame called 'flights_airports_merge'.
display(flights_airports_merge)
```

```python
# The next code carries out the count of the flight number by origin airport and destination airport.
Most_popular_routes = flights_airports_merge.groupby(['ORIGIN_AIRPORT','DESTINATION_AIRPORT']).apply(
                       lambda x: pd.Series({'nb_flights': len(x['FLIGHT_NUMBER'])})).reset_index()
display(Most_popular_routes)
```

```python
Most_popular_routes = Most_popular_routes.assign(rank_popular = lambda x: x.groupby(['ORIGIN_AIRPORT'])['nb_flights'].rank(method='first',
                                                                                                    ascending=False)
                          )[lambda x: x['rank_popular'] <= 3].reset_index(drop=True).drop(['nb_flights'], axis=1).assign(
    rank_popular= lambda x: 'Top' + x['rank_popular'].astype(int).map(str))
display(Most_popular_routes)
# The code above, it has ordered in descending order each category of 'ORIGIN_AIRPORT' by the number of flights.
# After that, the result has been filtered, taking in count the top 3 by origin airport.
# Finally, a new column called 'rank_popular has been added' which shows the word 'Top' plus the ranking position by 'origin airport'  
```

```python
Most_popular_routes.pivot_table(
    columns='rank_popular',
    values= 'DESTINATION_AIRPORT',
    index= 'ORIGIN_AIRPORT',
    aggfunc= np.sum
).rename_axis(None, axis=1).reset_index()
# The code above, we used the function pivot table to reorder the data frame. Finally, we obtain a table where 
# we can oberve more clearly the top 3 by origin airport. 
```
